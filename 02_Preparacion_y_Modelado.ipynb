## 9. Desarrollo y evaluación de modelos predictivos

### 9.1. Objetivo del modelado

Después de explorar los datos y revisar las variables disponibles, comenzaremos a desarrollar modelos de aprendizaje supervisado para estimar el riesgo de que un estudiante no ingrese a la educación superior durante los cinco años posteriores a la presentación de Saber 11.

Trabajaremos con la variable `enroll_5years`, donde 1 representa el ingreso a educación superior y 0 el no ingreso dentro del periodo establecido. Para interpretar los resultados como riesgo de no ingreso, tomaremos como referencia la probabilidad de la clase 0.

### 9.2. Modelos seleccionados

Probaremos tres algoritmos para comparar sus resultados:

- **Regresión logística:** servirá como modelo inicial de referencia. Permitirá evaluar la relación entre las características de los estudiantes y la probabilidad de ingreso.
- **Random Forest:** permitirá identificar relaciones más complejas entre las variables académicas, socioeconómicas y las características de los colegios.
- **XGBoost:** evaluaremos este algoritmo para comprobar si mejora las predicciones mediante la combinación de árboles de decisión.

### 9.3. Preparación y evaluación

Antes del entrenamiento revisaremos los datos faltantes, las variables seleccionadas y las cohortes con suficiente tiempo de seguimiento. Utilizaremos una separación temporal entre los datos de entrenamiento y prueba para evaluar el comportamiento de los modelos en estudiantes de cohortes posteriores.

Compararemos los resultados mediante métricas como ROC-AUC, precisión, recall, F1-score y calibración de probabilidades. También tendremos en cuenta el tiempo de entrenamiento y la facilidad para interpretar las predicciones.

Los experimentos se registrarán posteriormente en MLflow, con sus parámetros y resultados, para facilitar la comparación y seleccionar el modelo más adecuado para el prototipo.

### 9.4. Revisión de las cohortes y el periodo de seguimiento

Antes de entrenar los modelos, revisaremos los años de presentación de Saber 11 y las variables relacionadas con el ingreso a educación superior.

El análisis exploratorio mostró diferencias en las tasas de ingreso entre cohortes. Parte de estas diferencias está relacionada con el tiempo de seguimiento, ya que los estudiantes más recientes todavía no han completado los cinco años necesarios para evaluar la variable objetivo.

En esta etapa examinaremos la distribución de `enroll_5years` y la información disponible sobre el seguimiento de cada estudiante. Los resultados servirán para definir qué registros podremos incluir en el entrenamiento y la evaluación de los modelos.

### 9.4.2. Relación entre el tiempo registrado y el ingreso a educación superior

Se examinará la relación entre los valores de `ventana_sb11_es` y la clasificación de ingreso durante cinco años. Esta revisión busca identificar patrones en los registros y comprobar si la variable contiene información útil para entender cómo se construyó el objetivo.

El análisis se realizará inicialmente con la muestra disponible. Sus resultados servirán para orientar la revisión de la definición de las variables antes del entrenamiento.

### 9.4.4. Revisión del seguimiento de las cohortes

Para entrenar los modelos necesitamos identificar qué cohortes cuentan con información suficiente sobre el ingreso a educación superior.

Aunque la base incluye estudiantes que presentaron Saber 11 entre 2012 y 2025, los registros más recientes tienen menos tiempo de observación. Esto afecta la interpretación de `enroll_5years`, ya que algunos estudiantes todavía no han completado el periodo necesario para conocer su resultado.

Revisaremos las variables relacionadas con las fechas de presentación y el ingreso a educación superior. Con esta información buscaremos establecer una separación temporal adecuada para entrenar y evaluar los modelos, evitando clasificar como no ingreso casos cuyo seguimiento aún está incompleto.

```python
# Preparar y ejecutar el análisis correspondiente
_, meta = pyreadstat.read_dta(
    "data/Base de Datos (MIAD).dta",
    metadataonly=True
)

palabras_clave = [
    "año", "periodo", "ventana",
    "enroll", "year", "sem"
]

variables = pd.DataFrame({
    "Variable": meta.column_names,
    "Descripción": meta.column_labels
})

filtro = variables["Variable"].str.contains(
    "|".join(palabras_clave),
    case=False,
    regex=True,
    na=False
)

display(
    variables.loc[filtro].fillna("").style
    .set_caption("Variables relacionadas con el seguimiento educativo")
)
```

### 9.4.5. Identificación de cohortes con seguimiento suficiente

La revisión de las variables permitió identificar el año de presentación de Saber 11, el año de ingreso a educación superior y el tiempo transcurrido entre ambos eventos.

En este análisis compararemos las tasas de ingreso registradas para cada cohorte. Buscamos identificar los años con seguimiento incompleto, ya que esto afecta la variable `enroll_5years`.

Los resultados nos ayudarán a seleccionar los periodos de entrenamiento, validación y prueba, evitando clasificar como no ingreso a estudiantes que todavía no han completado los cinco años de observación.

```python
# Preparar y ejecutar el análisis correspondiente
columnas = [
    "año",
    "enroll_5years",
    "year_enrolled",
    "ventana_sb11_es"
]

resultados = []

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=columnas
)

for bloque, _ in lector:
    resumen_bloque = bloque.groupby("año").agg(
        total=("año", "size"),
        ingresaron=("enroll_5years", "sum"),
        sin_fecha_ingreso=("year_enrolled", lambda x: x.isna().sum()),
        sin_ventana=("ventana_sb11_es", lambda x: x.isna().sum())
    )
    resultados.append(resumen_bloque)

tabla_cohortes = (
    pd.concat(resultados)
    .groupby(level=0)
    .sum()
    .reset_index()
    .sort_values("año")
)

tabla_cohortes["no_ingresaron"] = (
    tabla_cohortes["total"] - tabla_cohortes["ingresaron"]
)

tabla_cohortes["tasa_ingreso_%"] = (
    100 * tabla_cohortes["ingresaron"] / tabla_cohortes["total"]
).round(2)

display(
    tabla_cohortes.style
    .format({
        "total": "{:,.0f}",
        "ingresaron": "{:,.0f}",
        "no_ingresaron": "{:,.0f}",
        "sin_fecha_ingreso": "{:,.0f}",
        "sin_ventana": "{:,.0f}",
        "tasa_ingreso_%": "{:.2f}%"
    })
    .set_caption("Distribución del ingreso a educación superior por cohorte")
)
```

Entre 2012 y 2019, los porcentajes registrados se mantienen aproximadamente entre el 52 % y el 56 %. En 2020 se observa una tasa del 61,94 %, mientras que en las cohortes posteriores los porcentajes disminuyen hasta llegar al 0 % en 2025.

Esta reducción no debe interpretarse directamente como una caída del acceso educativo. Las cohortes recientes tienen menos tiempo de seguimiento, lo que impide conocer todavía su resultado completo dentro de los cinco años posteriores a Saber 11.

Para el entrenamiento será necesario seleccionar cohortes con una ventana de observación suficiente y confirmar la fecha de corte de los registros de educación superior. Esta decisión busca evitar que el modelo aprenda clasificaciones afectadas por la falta de seguimiento.

### 9.5. Selección de cohortes para el entrenamiento

El siguiente paso será determinar qué años cuentan con seguimiento suficiente para construir una variable objetivo confiable.

Para ello, revisaremos la cobertura temporal de los registros de ingreso a educación superior y el comportamiento de `year_enrolled`. Una vez confirmada la fecha de corte de la información, definiremos los conjuntos

```python
# Preparar y ejecutar el análisis correspondiente
import pyreadstat
import pandas as pd
from collections import Counter
from IPython.display import display

conteo_ingresos = Counter()

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=["year_enrolled"]
)

for bloque, _ in lector:
    conteo_ingresos.update(
        bloque["year_enrolled"].dropna().value_counts().to_dict()
    )

tabla_ingresos = pd.DataFrame(
    sorted(conteo_ingresos.items()),
    columns=["Año de ingreso", "Registros"]
)

display(
    tabla_ingresos.style
    .format({"Registros": "{:,.0f}"})
    .set_caption("Distribución de los años de ingreso a educación superior")
)

print("Último año registrado:", tabla_ingresos["Año de ingreso"].max())
```

### 9.5.1. Cobertura temporal de los registros de ingreso

La revisión de `year_enrolled` permitió identificar registros de ingreso a educación superior entre 2012 y 2025. El último año disponible es 2025, con 704.488 registros asociados.

Estos valores representan los ingresos registrados en cada año y pueden incluir estudiantes que presentaron Saber 11 en cohortes anteriores. Por esta razón, no deben interpretarse directamente como tasas de ingreso de una cohorte específica.

Para evaluar el ingreso durante los cinco años posteriores a Saber 11, consideraremos inicialmente las cohortes de 2012 a 2020. Esta selección está sujeta a la confirmación de la cobertura completa de 2025 en la fuente original.

Las cohortes posteriores quedarán fuera del primer entrenamiento, debido a que todavía no cuentan con el mismo horizonte de observación.

### 9.5.2. Distribución de las cohortes candidatas para el modelado

Se revisarán las cohortes de 2012 a 2020 para conocer la cantidad de registros disponibles y la distribución de la variable `enroll_5years`.

Esta revisión permitirá establecer el tamaño inicial de los conjuntos de entrenamiento, validación y prueba. La selección definitiva dependerá de la cobertura temporal de los datos y de la revisión de posibles diferencias entre cohortes.

```python
# Preparar y ejecutar el análisis correspondiente
from IPython.display import display

# Selección inicial de cohortes
cohortes_modelo = tabla_cohortes[
    tabla_cohortes["año"].between(2012, 2020)
].copy()

# Distribución de la variable objetivo
cohortes_modelo["tasa_no_ingreso_%"] = (
    100 - cohortes_modelo["tasa_ingreso_%"]
).round(2)

# Resumen para el modelado
resumen_modelo = cohortes_modelo[
    [
        "año",
        "total",
        "ingresaron",
        "no_ingresaron",
        "tasa_ingreso_%",
        "tasa_no_ingreso_%"
    ]
].copy()

display(
    resumen_modelo.style
    .format({
        "total": "{:,.0f}",
        "ingresaron": "{:,.0f}",
        "no_ingresaron": "{:,.0f}",
        "tasa_ingreso_%": "{:.2f}%",
        "tasa_no_ingreso_%": "{:.2f}%"
    })
    .set_caption("Cohortes candidatas para el entrenamiento")
)

print(
    "Total de registros:",
    f"{resumen_modelo['total'].sum():,}"
)
```

### 9.5.3. Análisis de las cohortes seleccionadas

En 2020 se observa un aumento hasta el 61,94 %, resultado que requiere una revisión adicional para identificar posibles diferencias en la cobertura o composición de los datos.

La distribución de la variable objetivo muestra que contamos con registros de estudiantes que ingresaron y otros clasificados sin ingreso durante los cinco años posteriores a Saber 11.

Esta selección busca reducir los problemas de seguimiento presentes en las cohortes recientes. Antes de establecer la muestra definitiva, se verificará la cobertura temporal de los datos y se definirá una separación entre entrenamiento, validación y prueba.

Los tres modelos se evaluarán sobre las mismas particiones para facilitar la comparación de resultados.

### 9.6. División temporal de los datos

Se propone dividir los registros según el año de presentación de Saber 11, utilizando las cohortes de 2012 a 2017 para entrenamiento, 2018 y 2019 para validación y 2020 para prueba.

Esta separación permitirá evaluar los modelos con estudiantes de años posteriores a los utilizados durante el entrenamiento. La cohorte de prueba quedará sujeta a la confirmación de su seguimiento completo.

Se mantendrá la misma distribución para comparar la regresión logística, Random Forest y XGBoost bajo condiciones equivalentes.

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
from IPython.display import display

def asignar_conjunto(anio):
    if 2012 <= anio <= 2017:
        return "Entrenamiento"
    elif 2018 <= anio <= 2019:
        return "Validación"
    elif anio == 2020:
        return "Prueba"
    return "Excluido"

division = resumen_modelo.copy()

division["Conjunto"] = division["año"].apply(
    asignar_conjunto
)

resumen_division = (
    division.groupby("Conjunto")
    .agg(
        estudiantes=("total", "sum"),
        ingresaron=("ingresaron", "sum"),
        no_ingresaron=("no_ingresaron", "sum")
    )
    .reindex(["Entrenamiento", "Validación", "Prueba"])
)

resumen_division["Porcentaje"] = (
    100 * resumen_division["estudiantes"]
    / resumen_division["estudiantes"].sum()
).round(2)

display(
    resumen_division.style
    .format({
        "estudiantes": "{:,.0f}",
        "ingresaron": "{:,.0f}",
        "no_ingresaron": "{:,.0f}",
        "Porcentaje": "{:.2f}%"
    })
    .set_caption("Distribución temporal propuesta para los modelos")
)
```

### Continuación del análisis

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
from IPython.display import display

tabla_seleccion = tabla_cohortes[
    ["año", "total", "tasa_ingreso_%"]
].copy()

def clasificar_cohorte(anio):
    if anio <= 2017:
        return "Entrenamiento"
    elif anio <= 2019:
        return "Validación"
    elif anio == 2020:
        return "Prueba provisional"
    return "Excluida"

tabla_seleccion["Uso propuesto"] = (
    tabla_seleccion["año"].apply(clasificar_cohorte)
)

tabla_seleccion = tabla_seleccion.rename(
    columns={
        "año": "Año",
        "total": "Estudiantes",
        "tasa_ingreso_%": "Ingreso observado (%)"
    }
)

def destacar_filas(fila):
    if fila["Uso propuesto"] == "Excluida":
        return ["background-color: #FCE8E6"] * len(fila)
    elif fila["Uso propuesto"] == "Prueba provisional":
        return ["background-color: #FFF2D6"] * len(fila)
    return ["background-color: #EAF4EF"] * len(fila)

display(
    tabla_seleccion.style
    .apply(destacar_filas, axis=1)
    .format({
        "Estudiantes": "{:,.0f}",
        "Ingreso observado (%)": "{:.2f}%"
    })
    .set_caption(
        "Selección inicial de cohortes para el modelado"
    )
    .hide(axis="index")
)
```

### 9.7. Revisión de las variables predictoras

El siguiente paso consiste en revisar las características disponibles para construir los modelos de clasificación.

### 9.7.3. Disponibilidad de las variables según la cohorte

Se revisará la proporción de datos faltantes por año en las cuatro variables con mayores niveles de ausencia.

El propósito es identificar cambios en la disponibilidad de información entre cohortes. Estos resultados ayudarán a definir qué características conservar y cuáles requieren un tratamiento especial antes del entrenamiento.

El análisis se realizará sobre las cohortes candidatas, utilizando una lectura por bloques para controlar el consumo de memoria.

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
import pyreadstat
from IPython.display import display

variables_revision = [
    "estu_etnia",
    "estu_tieneetnia",
    "estu_inse_individual",
    "punt_global"
]

columnas = ["año"] + variables_revision

resultados = []

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=columnas
)

for bloque, _ in lector:
    bloque = bloque[
        bloque["año"].between(2012, 2020)
    ]

    if bloque.empty:
        continue

    conteo = bloque.groupby("año").size()
    faltantes = bloque.groupby("año")[
        variables_revision
    ].agg(lambda x: x.isna().sum())

    faltantes["total"] = conteo
    resultados.append(faltantes)

# Combinar los resultados de todos los bloques
resumen = (
    pd.concat(resultados)
    .groupby(level=0)
    .sum()
)

# Calcular porcentajes por cohorte
tabla_faltantes = (
    resumen[variables_revision]
    .div(resumen["total"], axis=0)
    .mul(100)
    .round(2)
)

display(
    tabla_faltantes.style
    .format("{:.2f}%")
    .background_gradient(
        cmap="Oranges",
        vmin=0,
        vmax=100
    )
    .set_caption(
        "Porcentaje de datos faltantes por cohorte"
    )
)
```

### 9.7.5. Verificación de valores ausentes en la base completa

La revisión inicial permitió identificar diferencias entre los porcentajes generales de datos faltantes y los resultados obtenidos por cohorte en algunas variables.

Para comprobar su origen, analizaremos todos los registros de `estu_etnia` y `estu_tieneetnia`, incluyendo valores nulos, cadenas vacías y categorías especiales.


```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
import pyreadstat
from collections import Counter
from IPython.display import display

variables_etnia = ["estu_etnia", "estu_tieneetnia"]

conteos = {col: Counter() for col in variables_etnia}
nulos = {col: 0 for col in variables_etnia}
vacios = {col: 0 for col in variables_etnia}
total = 0

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=variables_etnia
)

for bloque, _ in lector:
    total += len(bloque)

    for col in variables_etnia:
        serie = bloque[col].astype("string")
        limpia = serie.str.strip()

        nulos[col] += int(serie.isna().sum())
        vacios[col] += int(
            limpia.eq("").fillna(False).sum()
        )

        categorias = limpia.fillna("[NULO]")
        categorias = categorias.replace("", "[VACÍO]")

        conteos[col].update(
            categorias.value_counts().to_dict()
        )

resumen_etnia = pd.DataFrame([
    {
        "Variable": col,
        "Registros": total,
        "Nulos": nulos[col],
        "Vacíos": vacios[col],
        "Ausencia (%)": round(
            100 * (nulos[col] + vacios[col]) / total, 2
        )
    }
    for col in variables_etnia
])

display(resumen_etnia)

for col in variables_etnia:
    print(f"\nDistribución de {col}")

    tabla = pd.DataFrame(
        conteos[col].most_common(15),
        columns=["Categoría", "Cantidad"]
    )

    display(tabla)
```

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
import pyreadstat
from collections import Counter
from IPython.display import display

# Contadores para cada variable categórica
conteos_cat = {
    col: Counter() for col in variables_categoricas
}

total_registros = 0

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=["año"] + variables_categoricas
)

for bloque, _ in lector:

    # Cohortes candidatas para el modelado
    bloque = bloque[bloque["año"].between(2012, 2020)]

    total_registros += len(bloque)

    for col in variables_categoricas:

        valores = (
            bloque[col]
            .astype("string")
            .str.strip()
            .fillna("[SIN DATO]")
            .replace("", "[SIN DATO]")
        )

        conteos_cat[col].update(
            valores.value_counts().to_dict()
        )

# Resumen de las variables categóricas
resumen_categorias = []

for col in variables_categoricas:

    contador = conteos_cat[col]

    resumen_categorias.append({
        "Variable": col,
        "Categorías distintas": len(contador),
        "Sin información": contador.get("[SIN DATO]", 0),
        "Sin información (%)": round(
            100 * contador.get("[SIN DATO]", 0)
            / total_registros, 2
        ),
        "Categoría más frecuente": contador.most_common(1)[0][0]
    })

tabla_categorias = pd.DataFrame(resumen_categorias)

display(
    tabla_categorias.style
    .hide(axis="index")
    .format({
        "Sin información": "{:,.0f}",
        "Sin información (%)": "{:.2f}%"
    })
    .set_caption(
        "Revisión de categorías en las cohortes 2012 a 2020"
    )
)

print("Registros analizados:", f"{total_registros:,}")
```

### 9.9.7. Preparación del conjunto de datos para la regresión logística

Para comenzar el entrenamiento construiremos una muestra de las cohortes seleccionadas, manteniendo separados los registros de entrenamiento y validación.

La primera prueba servirá para comprobar que el procesamiento de las variables funciona correctamente antes de trabajar con un volumen mayor de información.

Las variables numéricas se estandarizarán y los valores faltantes se tratarán mediante imputación. Las características categóricas se transformarán con One-Hot Encoding, conservando las categorías especiales identificadas durante la exploración.

La variable objetivo se ajustará para que el valor 1 represente el riesgo de no ingreso a educación superior dentro de los cinco años posteriores a Saber 11.

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
import pyreadstat
import numpy as np
from IPython.display import display

# Columnas necesarias
columnas_modelado = (
    ["año", "enroll_5years"] + variables_modelo
)

muestras = []
semilla = 42

lector = pyreadstat.read_file_in_chunks(
    pyreadstat.read_dta,
    "data/Base de Datos (MIAD).dta",
    chunksize=100000,
    usecols=columnas_modelado
)

for bloque, _ in lector:

    # Seleccionar cohortes de entrenamiento y validación
    bloque = bloque[
        bloque["año"].between(2012, 2019)
    ]

    if bloque.empty:
        continue

    # Selección reproducible del 5 %
    muestra_bloque = bloque.sample(
        frac=0.05,
        random_state=semilla
    )

    muestras.append(muestra_bloque)

# Unificar las muestras
datos_modelo = pd.concat(
    muestras,
    ignore_index=True
)

# Crear variable de riesgo de no ingreso
datos_modelo["riesgo_no_ingreso"] = (
    1 - datos_modelo["enroll_5years"]
)

# Separar conjuntos temporalmente
train_df = datos_modelo[
    datos_modelo["año"].between(2012, 2017)
].copy()

val_df = datos_modelo[
    datos_modelo["año"].between(2018, 2019)
].copy()

resumen_muestra = pd.DataFrame({
    "Conjunto": ["Entrenamiento", "Validación"],
    "Registros": [len(train_df), len(val_df)],
    "Riesgo promedio (%)": [
        train_df["riesgo_no_ingreso"].mean() * 100,
        val_df["riesgo_no_ingreso"].mean() * 100
    ]
})

display(
    resumen_muestra.style
    .hide(axis="index")
    .format({
        "Registros": "{:,.0f}",
        "Riesgo promedio (%)": "{:.2f}%"
    })
    .set_caption(
        "Muestra inicial para la regresión logística"
    )
)

```

### 9.10. Entrenamiento del primer modelo: regresión logística

Usaremos las 19 variables seleccionadas anteriormente. Para prepararlas, aplicaremos imputación a los valores numéricos faltantes, estandarización y codificación de las categorías.

Estas transformaciones se ajustarán únicamente con los datos de entrenamiento. Después se aplicarán al conjunto de validación sin modificar los parámetros aprendidos.

Evaluaremos el modelo mediante ROC-AUC, precisión, recall, F1-score y exactitud. Así podremos conocer su capacidad inicial para identificar estudiantes con riesgo de no ingreso y comparar sus resultados con los próximos algoritmos.

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from IPython.display import display

# Preparar variables de entrenamiento y validación
X_train = train_df[variables_modelo].copy()
X_val = val_df[variables_modelo].copy()

y_train = train_df["riesgo_no_ingreso"].astype(int)
y_val = val_df["riesgo_no_ingreso"].astype(int)

# Convertir variables numéricas y valores no válidos
for columna in variables_numericas:
    X_train[columna] = pd.to_numeric(
        X_train[columna], errors="coerce"
    )
    X_val[columna] = pd.to_numeric(
        X_val[columna], errors="coerce"
    )

# Normalizar valores categóricos
for columna in variables_categoricas:
    for datos in [X_train, X_val]:
        datos[columna] = (
            datos[columna]
            .astype("string")
            .str.strip()
            .replace("", pd.NA)
            .fillna("Sin dato")
            .astype(str)
        )

# Estrato 0 como categoría independiente
for datos in [X_train, X_val]:
    datos["fami_estratovivienda"] = (
        datos["fami_estratovivienda"]
        .replace({"0": "Sin estrato"})
    )

# Transformaciones numéricas
procesamiento_numerico = Pipeline([
    ("imputador", SimpleImputer(strategy="median")),
    ("escalador", StandardScaler())
])

# Transformaciones categóricas
procesamiento_categorico = Pipeline([
    ("codificador", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

# Unir transformaciones
preprocesador = ColumnTransformer([
    ("numericas", procesamiento_numerico,
     variables_numericas),
    ("categoricas", procesamiento_categorico,
     variables_categoricas)
])

# Construir el modelo completo
modelo_logistico = Pipeline([
    ("preprocesamiento", preprocesador),
    ("clasificador", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Entrenar
modelo_logistico.fit(X_train, y_train)

# Predecir en validación
predicciones = modelo_logistico.predict(X_val)

probabilidades = modelo_logistico.predict_proba(
    X_val
)[:, 1]

# Evaluar
resultados_logisticos = pd.DataFrame({
    "Métrica": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "ROC-AUC"
    ],
    "Resultado": [
        accuracy_score(y_val, predicciones),
        precision_score(y_val, predicciones),
        recall_score(y_val, predicciones),
        f1_score(y_val, predicciones),
        roc_auc_score(y_val, probabilidades)
    ]
})

print("Regresión logística entrenada correctamente")

display(
    resultados_logisticos.style
    .hide(axis="index")
    .format({"Resultado": "{:.4f}"})
    .set_caption(
        "Evaluación inicial de la regresión logística"
    )
)
```

## 9.14. Evaluación de umbrales de clasificación

Después de comparar los tres modelos, revisaremos cómo cambian sus resultados al modificar el umbral utilizado para clasificar a los estudiantes.

Hasta ahora hemos trabajado con un umbral de 0,50. Sin embargo, este valor no necesariamente es el más adecuado para identificar estudiantes con riesgo de no ingresar a educación superior.

Evaluaremos diferentes umbrales para observar los cambios en precisión, recall y F1-score. También revisaremos cuántos estudiantes con no ingreso quedarían sin detectar en cada escenario.

Este análisis permitirá conocer el equilibrio entre identificar más estudiantes en riesgo y evitar clasificar incorrectamente a quienes sí ingresarán.

```python
# Preparar y ejecutar el análisis correspondiente
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from IPython.display import display

modelos_probabilidades = {
    "Regresión logística": probabilidades,
    "Random Forest": probabilidades_rf,
    "XGBoost": probabilidades_xgb
}

umbrales = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70]

resultados_umbrales = []

for nombre, probs in modelos_probabilidades.items():

    for umbral in umbrales:

        pred = (probs >= umbral).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val, pred, labels=[0, 1]
        ).ravel()

        resultados_umbrales.append({
            "Modelo": nombre,
            "Umbral": umbral,
            "Precision": precision_score(
                y_val, pred, zero_division=0
            ),
            "Recall": recall_score(
                y_val, pred, zero_division=0
            ),
            "F1-score": f1_score(
                y_val, pred, zero_division=0
            ),
            "No ingresos sin detectar": fn,
            "Alertas generadas": tp + fp
        })

tabla_umbrales = pd.DataFrame(resultados_umbrales)

display(
    tabla_umbrales.style
    .hide(axis="index")
    .format({
        "Umbral": "{:.2f}",
        "Precision": "{:.3f}",
        "Recall": "{:.3f}",
        "F1-score": "{:.3f}",
        "No ingresos sin detectar": "{:,.0f}",
        "Alertas generadas": "{:,.0f}"
    })
    .set_caption(
        "Comparación de umbrales de clasificación"
    )
)
```

### 9.14.1. Resultados de la evaluación de umbrales

La comparación muestra que los tres modelos aumentan su capacidad de identificar estudiantes con riesgo de no ingreso cuando reducimos el umbral de clasificación.

En XGBoost, el recall pasa del 64,03 % con un umbral de 0,50 al 80,6 % cuando utilizamos 0,35. Esto permite identificar 3.781 casos adicionales de no ingreso dentro del conjunto de validación.

El cambio también afecta la precisión, que disminuye del 67,62 % al 59,6 %. Esto significa que aumentan las alertas sobre estudiantes que finalmente sí ingresaron a educación superior.

Entre los valores evaluados, XGBoost alcanza su mayor F1-score con un umbral de 0,35, mientras que la regresión logística presenta un resultado cercano. Por ahora, consideraremos este valor como una alternativa para las siguientes pruebas.

La elección definitiva dependerá de la cantidad de estudiantes que una institución pueda atender y del equilibrio entre identificar más casos de riesgo y generar alertas innecesarias.

### 9.14.2. Relación entre las alertas generadas y los casos detectados

Para entender el efecto de los umbrales, compararemos la cantidad de alertas generadas con el recall obtenido por cada algoritmo.

Esta relación permitirá observar cuántos estudiantes debemos identificar como población de riesgo para detectar una mayor proporción de los casos reales de no ingreso.

El análisis servirá como referencia para seleccionar un umbral de clasificación según la capacidad de acompañamiento de una institución educativa.

```python
# Preparar y ejecutar el análisis correspondiente
import matplotlib.pyplot as plt
from pathlib import Path

colores = {
    "Regresión logística": "#17365D",
    "Random Forest": "#E58A36",
    "XGBoost": "#168A80"
}

fig, ax = plt.subplots(figsize=(12, 6.5))
fig.patch.set_facecolor("white")

for modelo, color in colores.items():

    datos = tabla_umbrales[
        tabla_umbrales["Modelo"] == modelo
    ].sort_values("Alertas generadas")

    ax.plot(
        datos["Alertas generadas"],
        datos["Recall"] * 100,
        marker="o",
        markersize=6,
        linewidth=2.5,
        color=color,
        label=modelo
    )

# Resaltar XGBoost con umbral 0.35
punto = tabla_umbrales[
    (tabla_umbrales["Modelo"] == "XGBoost") &
    (tabla_umbrales["Umbral"] == 0.35)
].iloc[0]

ax.scatter(
    punto["Alertas generadas"],
    punto["Recall"] * 100,
    s=180,
    color=colores["XGBoost"],
    edgecolors="white",
    linewidths=2,
    zorder=5
)

ax.annotate(
    "XGBoost | Umbral 0,35\nRecall: 80,6 %",
    xy=(
        punto["Alertas generadas"],
        punto["Recall"] * 100
    ),
    xytext=(-125, -60),
    textcoords="offset points",
    fontsize=10,
    fontweight="bold",
    color=colores["XGBoost"],
    arrowprops=dict(
        arrowstyle="->",
        color=colores["XGBoost"]
    )
)

ax.set_title(
    "Detección de estudiantes en riesgo según alertas generadas",
    fontsize=17,
    fontweight="bold",
    color="#17365D",
    pad=20
)

ax.set_xlabel(
    "Cantidad de estudiantes identificados en riesgo",
    fontsize=11
)

ax.set_ylabel(
    "Casos reales de no ingreso detectados (Recall %)",
    fontsize=11
)

ax.grid(alpha=0.15)
ax.legend(frameon=False, loc="upper left")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

ruta = Path("reports/modelos")
ruta.mkdir(parents=True, exist_ok=True)

plt.savefig(
    ruta / "08_comparacion_umbrales_alertas.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

La gráfica muestra una relación similar entre la cantidad de alertas y el recall de los tres modelos. A medida que identificamos más estudiantes como población de riesgo, aumenta la proporción de casos reales de no ingreso que logramos detectar.

Las diferencias entre los modelos son pequeñas. Esto indica que, para una cantidad parecida de alertas, los tres modelos ofrecen resultados similares en esta primera evaluación.

En XGBoost, el umbral de 0,35 genera 30.877 alertas y permite identificar el 80,6 % de los estudiantes que no ingresaron. Con esta configuración quedan sin detectar 4.431 casos, frente a los 8.212 registrados cuando utilizamos el umbral de 0,50.

Este resultado muestra que ajustar el umbral puede mejorar la detección sin modificar el entrenamiento. Sin embargo, una mayor cantidad de alertas también implica más estudiantes que necesitarían una revisión o acompañamiento.

Por ahora, mantendremos el umbral de 0,35 como candidato para XGBoost.

## 9.15. Evaluación de la calibración de probabilidades

Después de revisar las métricas y los umbrales, evaluaremos la calibración de las probabilidades estimadas por cada modelo.

Una buena calibración significa que las probabilidades predichas se aproximan a las proporciones observadas. Por ejemplo, entre estudiantes a quienes el modelo asigna un riesgo cercano al 70 %, esperaríamos encontrar una frecuencia de no ingreso cercana a ese valor.

Este análisis es relevante para el tablero del proyecto, ya que buscamos presentar estimaciones de riesgo que puedan interpretarse y apoyar la identificación de estudiantes que requieren acompañamiento.

Compararemos la regresión logística, Random Forest y XGBoost mediante curvas de calibración y el Score. Los resultados permitirán identificar si alguno de los modelos tiende a sobreestimar o subestimar el riesgo.

```python
# Preparar y ejecutar el análisis correspondiente
import matplotlib.pyplot as plt
from pathlib import Path

from sklearn.calibration import calibration_curve
from sklearn.metrics import brier_score_loss

modelos_calibracion = {
    "Regresión logística": (
        probabilidades, "#17365D"
    ),
    "Random Forest": (
        probabilidades_rf, "#E58A36"
    ),
    "XGBoost": (
        probabilidades_xgb, "#168A80"
    )
}

fig, ax = plt.subplots(figsize=(10, 7))

resultados_brier = {}

for nombre, (probs, color) in modelos_calibracion.items():

    fraccion_positivos, prob_media = calibration_curve(
        y_val,
        probs,
        n_bins=10,
        strategy="quantile"
    )

    brier = brier_score_loss(y_val, probs)
    resultados_brier[nombre] = brier

    ax.plot(
        prob_media,
        fraccion_positivos,
        marker="o",
        markersize=7,
        linewidth=2.5,
        color=color,
        label=f"{nombre} (Brier = {brier:.4f})"
    )

ax.plot(
    [0, 1], [0, 1],
    linestyle="--",
    color="#94A3B8",
    linewidth=1.5,
    label="Calibración ideal"
)

ax.set_title(
    "Calibración de probabilidades de riesgo",
    fontsize=18,
    fontweight="bold",
    color="#17365D",
    pad=20
)

ax.set_xlabel("Probabilidad promedio estimada")
ax.set_ylabel("Proporción real de no ingreso")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(alpha=0.15)
ax.legend(loc="upper left", frameon=False)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

ruta = Path("reports/modelos")
ruta.mkdir(parents=True, exist_ok=True)

plt.savefig(
    ruta / "09_calibracion_modelos.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print("Brier Score por modelo:")
for nombre, valor in resultados_brier.items():
    print(f"{nombre}: {valor:.4f}")
```

### 9.15.1. Resultados de la calibración de probabilidades

La evaluación muestra que los tres modelos presentan curvas cercanas a la línea de calibración ideal. Esto indica una correspondencia razonable entre las probabilidades estimadas y las proporciones observadas de no ingreso.

XGBoost obtuvo el menor Score, con 0,1967, seguido de Random Forest con 0,1985 y la regresión logística con 0,1987. Las diferencias son pequeñas, aunque XGBoost presenta el menor error promedio de probabilidad en esta evaluación.

Al revisar las curvas también encontramos ligeras diferencias en algunos grupos de riesgo. Por esta razón, todavía no podemos asumir que las probabilidades sean igual de confiables para todos los estudiantes.

Estos resultados son relevantes para el proyecto, donde buscamos mostrar estimaciones de riesgo que ayuden a identificar estudiantes que podrían necesitar acompañamiento.

Por ahora, XGBoost se mantiene como candidato principal debido a sus resultados en ROC-AUC, F1-score y Brier Score. Antes de elegirlo revisaremos si su desempeño mejora con otros parámetros y si mantiene resultados similares en una evaluación independiente.
